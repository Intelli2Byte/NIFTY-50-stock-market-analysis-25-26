```python
from scipy import stats

# Create a copy to preserve original data
import pandas as pd

df = pd.read_csv("NIFTY50.csv")  # Replace with your file name

df_clean = df.copy()

technical_cols = [
    'SMA_5', 'SMA_10', 'SMA_20', 'SMA_50',
    'EMA_5', 'EMA_10', 'EMA_20', 'EMA_50',
    'Volatility_5D', 'Volatility_10D', 'Volatility_20D',
    'ATR_14', 'RSI_14', 'RSI_Signal',
    'MACD', 'MACD_Signal', 'MACD_Histogram',
    'BB_Upper', 'BB_Mid', 'BB_Lower', 'BB_Width', 'BB_Position'
]

for col in technical_cols:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].ffill().bfill()
```

```python
# B. Volume indicators
volume_cols = ['Volume_MA_20', 'Volume_Ratio', 'Turnover_MA_20']
for col in volume_cols:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].fillna(method='ffill').fillna(method='bfill')

# C. Return calculations (use 0 for first day)
return_cols = ['Daily_Return_Pct', 'Change_from_Prev_Close', 'Gap_Pct', 'Gap_Up_Down']
for col in return_cols:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].fillna(0)

# D. Categorical columns (MACD_Crossover, Trend)
if 'MACD_Crossover' in df_clean.columns:
    df_clean['MACD_Crossover'] = df_clean['MACD_Crossover'].fillna('None')

if 'Trend' in df_clean.columns:
    df_clean['Trend'] = df_clean['Trend'].fillna(method='ffill').fillna('Neutral')

# Verify missing values are handled
print("\n✅ Missing values after cleaning:")
remaining_missing = df_clean.isnull().sum()
print(remaining_missing[remaining_missing > 0])
if remaining_missing.sum() == 0:
    print("All missing values successfully handled!")
```

```python
# A. Price Movement Analysis
print("\n📊 PRICE MOVEMENT SUMMARY")
print("-" * 80)
price_summary = {
    'Starting Price': df_clean['Close'].iloc[0],
    'Ending Price': df_clean['Close'].iloc[-1],
    'Total Return (%)': ((df_clean['Close'].iloc[-1] / df_clean['Close'].iloc[0]) - 1) * 100,
    'Highest Price': df_clean['High'].max(),
    'Lowest Price': df_clean['Low'].min(),
    'Average Close': df_clean['Close'].mean(),
    'Price Volatility (Std)': df_clean['Close'].std(),
    'Max Daily Gain (%)': df_clean['Daily_Return_Pct'].max(),
    'Max Daily Loss (%)': df_clean['Daily_Return_Pct'].min(),
    'Max Drawdown (%)': df_clean['Drawdown_Pct'].min()
}

for key, value in price_summary.items():
    print(f"{key:.<40} {value:>15,.2f}")
```

```python
# B. Trading Activity Analysis
print("\n📈 TRADING ACTIVITY SUMMARY")
print("-" * 80)
trading_summary = {
    'Total Trading Days': len(df_clean),
    'Up Days': df_clean['Up_Day'].sum(),
    'Down Days': len(df_clean) - df_clean['Up_Day'].sum(),
    'Win Rate (%)': (df_clean['Up_Day'].sum() / len(df_clean)) * 100,
    'Avg Daily Volume (M)': df_clean['Shares_Traded'].mean() / 1_000_000,
    'Max Volume (M)': df_clean['Shares_Traded'].max() / 1_000_000,
    'Avg Turnover (Cr)': df_clean['Turnover_Cr'].mean(),
    'Total Turnover (Cr)': df_clean['Turnover_Cr'].sum()
}

for key, value in trading_summary.items():
    print(f"{key:.<40} {value:>15,.2f}")
```

```python
# C. Technical Indicator Analysis
print("\n🔧 TECHNICAL INDICATORS SUMMARY")
print("-" * 80)
tech_summary = {
    'Current RSI': df_clean['RSI_14'].iloc[-1],
    'Avg RSI': df_clean['RSI_14'].mean(),
    'Overbought Days (RSI>70)': (df_clean['RSI_14'] > 70).sum(),
    'Oversold Days (RSI<30)': (df_clean['RSI_14'] < 30).sum(),
    'Current MACD': df_clean['MACD'].iloc[-1],
    'Avg Volatility (20D)': df_clean['Volatility_20D'].mean(),
    'Current ATR': df_clean['ATR_14'].iloc[-1]
}

for key, value in tech_summary.items():
    print(f"{key:.<40} {value:>15,.2f}")

# D. Trend Analysis
print("\n📉 TREND DISTRIBUTION")
print("-" * 80)
trend_counts = df_clean['Trend'].value_counts()
for trend, count in trend_counts.items():
    pct = (count / len(df_clean)) * 100
    print(f"{trend:.<40} {count:>5} ({pct:>5.1f}%)")
```

```python
# ============================================================================
# 4. CORRELATION ANALYSIS
# ============================================================================

print("\n" + "="*80)
print("CORRELATION ANALYSIS")
print("="*80)

# Select key numerical columns for correlation
corr_cols = ['Close', 'Daily_Return_Pct', 'Shares_Traded', 'Turnover_Cr',
             'RSI_14', 'MACD', 'Volatility_20D', 'ATR_14', 'Volume_Ratio']

correlation_matrix = df_clean[corr_cols].corr()

plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='coolwarm', 
            center=0, square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlation Matrix - Key Indicators', fontsize=16, fontweight='bold', pad=20)
plt.tight_layout()
plt.savefig('correlation_matrix.png', dpi=300, bbox_inches='tight')
print("✅ Correlation matrix saved as 'correlation_matrix.png'")
plt.show()
```

```python
# ============================================================================
# 5. STATISTICAL TESTS
# ============================================================================

print("\n" + "="*80)
print("STATISTICAL TESTS")
print("="*80)

# Normality test for returns
statistic, p_value = stats.normaltest(df_clean['Daily_Return_Pct'].dropna())
print(f"\n📊 Normality Test (Daily Returns):")
print(f"   Statistic: {statistic:.4f}")
print(f"   P-value: {p_value:.4f}")
print(f"   Result: {'Returns are NOT normally distributed' if p_value < 0.05 else 'Returns appear normally distributed'}")
```

```python
# Skewness and Kurtosis
skewness = df_clean['Daily_Return_Pct'].skew()
kurtosis = df_clean['Daily_Return_Pct'].kurtosis()
print(f"\n📈 Distribution Characteristics:")
print(f"   Skewness: {skewness:.4f} ({'Negative (left-tailed)' if skewness < 0 else 'Positive (right-tailed)'})")
print(f"   Kurtosis: {kurtosis:.4f} ({'Leptokurtic (fat tails)' if kurtosis > 0 else 'Platykurtic (thin tails)'})")
```
